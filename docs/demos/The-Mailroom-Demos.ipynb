## Seed demo traces (into Langfuse)

Demo runs are written **into** Langfuse (`env demo`). The visualizer still
reads Langfuse only.

```bash
python scripts/seed_demo.py
python scripts/seed_demo.py --list-scenarios
python scripts/seed_demo.py --check --check-api
```

The pixel `D` key does not fabricate envelopes on a live floor (`?demo=1` only,
and only when the source is down).


````python
from pathlib import Path

root = Path.cwd()
if (root / "docs" / "screenshots").is_dir():
    shots = root / "docs" / "screenshots"
    demos = root / "docs" / "demos"
elif (root / "screenshots").is_dir():
    shots = root / "screenshots"
    demos = root
else:
    raise SystemExit(f"cannot find screenshots from {root}")

pngs = sorted(p.name for p in shots.glob("*.png"))
video = demos / "tui-server-observatory-desk-walkthrough.mp4"
print(f"{len(pngs)} stills in {shots}")
print("video", "OK" if video.exists() else "MISSING", video.name, video.stat().st_size if video.exists() else 0)
for name in pngs:
    print(" ", name)
assert video.exists(), "walkthrough mp4 missing from docs/demos/"
assert len(pngs) >= 17, pngs
````
